````python
with open("rosalind_ini6.txt", "r") as f:
  line = f.readline()

words = line.split()
word_count = {}
for word in words:
  if word not in word_count:
    word_count[word] = words.count(word)

for key, value in word_count.items():
  print(key, value)
````

####**TASK P23.2** <font color=orange> **Complete Rosalind #9  Computing GC Content**</font>

````python
from Bio.SeqIO import parse
from Bio.SeqUtils import gc_fraction

max_gc = 0
max_id = ""

for seq_record in parse("rosalind_gc_1224_1_dataset.txt", "fasta"):
  current_gc = gc_fraction(seq_record.seq) * 100
  if current_gc > max_gc:
    max_gc = current_gc
    max_id = seq_record.id
  else:
    pass

print(max_id)
print(max_gc)
````
